## Загрузить ZIP
Выбери `Witch_tetris_ANDROID_TEST_v6_17_touch.zip`.


```python
from google.colab import files
uploaded = files.upload()
zip_name = next(iter(uploaded))
print('Uploaded:', zip_name)
```

```python
import os, shutil, zipfile, pathlib
root = pathlib.Path('/content/witch_android_test')
if root.exists(): shutil.rmtree(root)
root.mkdir()
with zipfile.ZipFile('/content/' + zip_name) as z: z.extractall(root)
# ZIP may contain files directly or one top-level folder. Find buildozer.spec.
specs = list(root.rglob('buildozer.spec'))
assert specs, 'buildozer.spec not found'
project = specs[0].parent
os.chdir(project)
print('Project:', project)
print(open('buildozer.spec', encoding='utf-8').read())
```
